```python
import wandb
import pandas as pd
import plotly.express as px

# 1) Read your Optuna‐XGBoost trials CSV
df = pd.read_csv("data/processed/xgboost_results.csv")

# 2) Start a new W&B run
run = wandb.init(
    project="Phase1_trial_plots",
    name="xgb_parcoords",
    config={"notes": "XGBoost Optuna sweep, phase1"}
)

# 3) Build a Plotly parallel‐coordinates figure
#    - color by your metric column (e.g. 'value' = validation RMSE)
#    - set up axes for each hyperparameter + the metric
fig = px.parallel_coordinates(
    df,
    color="value",  # your metric (lower is better, so you can invert the scale below)
    dimensions=[
        "params_learning_rate",
        "params_max_depth",
        "params_subsample",
        "params_colsample_bytree",
        "value"
    ],
    color_continuous_scale=px.colors.sequential.Viridis_r,  # invert so darker=better
    labels={
        "params_learning_rate": "Learning Rate",
        "params_max_depth":      "Max Depth",
        "params_subsample":      "Subsample",
        "params_colsample_bytree":"Col Sample",
        "value":                 "Val RMSE"
    },
    title="XGBoost Optuna Trials: Hyperparams ↔ Val RMSE"
)

# 4) Log the Plotly figure
wandb.log({"optuna_parallel_coordinates": fig})

# 5) Finish
run.finish()
```

```python
import wandb
import pandas as pd
import plotly.express as px

# 1) Read your Optuna MLP trials CSV
df = pd.read_csv("data/processed/optuna_summary.csv")

# 2) Start a new W&B run
run = wandb.init(
    project="Phase1_trial_plots",
    name="mlp_parcoords",
    config={"notes": "MLP Optuna sweep, phase1"}
)

# 3) Build a Plotly parallel‐coordinates figure
fig = px.parallel_coordinates(
    df,
    color="score",   # your validation RMSE
    dimensions=[
        "n_layers",
        "units",
        "dropout",
        "weight_decay",  # L2
        "lr",            # learning rate
        "score"
    ],
    color_continuous_scale=px.colors.sequential.Viridis_r,  # darker lines = lower RMSE
    labels={
        "n_layers":      "Num Layers",
        "units":         "Units/Layer",
        "dropout":       "Dropout",
        "weight_decay":  "Weight Decay",
        "lr":            "Learning Rate",
        "score":         "Val RMSE"
    },
    title="MLP Optuna Trials: Hyperparams ↔ Validation RMSE"
)

# 4) Log the Plotly figure to W&B
wandb.log({"mlp_optuna_parallel_coordinates": fig})

# 5) Finish the run
run.finish()
```

```python
import wandb
import pandas as pd
import plotly.express as px

# 1) Read your CatBoost Optuna trials CSV
df = pd.read_csv(
    "data/processed/catboost_trials.csv",
    sep=";",                 # <— tell pandas to use semicolons
    engine="python"          # more forgiving parser
)

# 2) Start a new W&B run
run = wandb.init(
    project="Phase1_trial_plots",
    name="catboost_parcoords",
    config={"notes": "CatBoost Optuna sweep, phase1"}
)

# 3) Build a Plotly parallel-coordinates figure
fig = px.parallel_coordinates(
    df,
    color="value",    # your validation RMSE column
    dimensions=[
        "learning_rate",    # column from CSV
        "depth",
        "subsample",
        "colsample_bylevel",
        "value"
    ],
    color_continuous_scale=px.colors.sequential.Viridis_r,
    labels={
        "learning_rate":     "Learning Rate",
        "depth":             "Depth",
        "subsample":         "Subsample",
        "colsample_bylevel": "Col Sample",
        "value":             "Val RMSE"
    },
    title="CatBoost Optuna Trials: Hyperparams ↔ Validation RMSE"
)

# 4) Log the Plotly figure to W&B
wandb.log({"catboost_optuna_parallel_coordinates": fig})

# 5) Finish the run
run.finish()
```
