# S

### Import libraries

```python
import pandas as pd       # Data manipulation  
import numpy as np        # Numerical operations  
import matplotlib.pyplot as plt  # Plotting  
```

### Load & inspect the dataset

```python
path = 'input/'
df = pd.read_csv(path + 'insurance_classified.csv')
print('\nNumber of rows and columns in the data set: ', df.shape)
print()
print(df.head())
```

### Plot the “charges_cat” distribution

```python
df["charges_cat"].value_counts().sort_index().plot.bar(rot=0, grid=True)
plt.xlabel("Charges category")
plt.ylabel("Number of persons")
plt.show()
```

### Impute & clean the data

```python
from sklearn.impute import SimpleImputer

# 1) Fill any missing cells with the column’s most frequent value
imp = SimpleImputer(missing_values=np.nan, strategy='most_frequent')
X = imp.fit_transform(df)
df_tr = pd.DataFrame(X, columns=df.columns, index=df.index)

# 2) Remove any unrealistic ages and drop the original charges
df_tr = df_tr[df_tr["age"].astype(float) < 100]
df_tr = df_tr.drop("charges", axis=1)

# 3) Quick schema check
df_tr.info()
```

Replacing NaNs with the modal value in each column

Filtering out any rows where age ≥ 100

Dropping the raw charges column

### Split into training and test sets

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    df_tr.drop('charges_cat', axis=1),
    df_tr['charges_cat'],
    test_size=0.3,
    random_state=42
)
```

holding out 30% of the data for testing, and keeping 70% for training, stratified by default randomness (via random_state)

### Scale all features to zero mean & unit variance

```python
from sklearn.preprocessing import StandardScaler

std_scaler = StandardScaler()

# fit on train, transform both train & test
X_train_scaled = std_scaler.fit_transform(X_train)
X_train_scaled = pd.DataFrame(
    X_train_scaled,
    columns=["age", "sex", "bmi", "children", "smoker", "region"],
    index=X_train.index
)
X_train_scaled.info()

X_test_scaled = std_scaler.transform(X_test)
X_test_scaled = pd.DataFrame(
    X_test_scaled,
    columns=["age", "sex", "bmi", "children", "smoker", "region"],
    index=X_test.index
)
```

fitting a StandardScaler on the training features, then applying the same scaling to the test set so that all numeric & encoded columns share the same scale.

### Train & evaluate a supervised SVM classifier

```python
from sklearn.svm import SVC
from sklearn.model_selection import cross_val_score
from sklearn.metrics import accuracy_score

# 1) Fit an SVM on the scaled training data
model = SVC()
model.fit(X_train_scaled, y_train)

# 2) Cross-validate on the training set
scores = cross_val_score(model, X_train_scaled, y_train, cv=3, scoring="accuracy")
print("Cross-validated accuracy (train):", scores.mean(), "±", scores.std())

# 3) Evaluate on the held-out test set
y_pred = model.predict(X_test_scaled)
test_acc = accuracy_score(y_test, y_pred)
print("Test accuracy (SVM):", test_acc)
```

Fitting an SVM with default settings

Checking its 3-fold cross-validation performance on the training data

Reporting final accuracy on the unseen test set.

### Randomly mask 30% of the training labels

```python
rng = np.random.RandomState(2)
random_unlabeled_points = rng.rand(y_train.shape[0]) < 0.3   # 30% unlabeled
y_train[random_unlabeled_points] = -1                         # -1 marks “unknown”
```

we simulate a semi-supervised scenario by hiding (setting to –1) about 30% of the true charges_cat labels in the training set

### Quick sanity-check of the training split

```python
print(X_train_scaled)
print(y_train)
```

### Separate labeled and unlabeled portions

```python
# Labeled subset
X_train_labeled   = X_train_scaled[y_train != -1]
y_train_labeled   = y_train[y_train != -1]

# Unlabeled subset
X_train_unlabeled = X_train_scaled[y_train == -1]
y_train_unlabeled = y_train[y_train == -1]
```

splitting the training data into the part with known labels (for supervised seeding) and the part without (to be inferred by our semi-supervised model)

### Dimensionality reduction with PCA

```python
from sklearn.decomposition import PCA

# 1) Compute PCA just on the labeled training rows
X_labeled = X_train_scaled[y_train != -1]     # 647×6
y_labeled = y_train[y_train != -1]            # length 647

pca = PCA(n_components=2, random_state=42)
X_pca_labeled = pca.fit_transform(X_labeled)  # 647×2

# 2) Scatter‐plot PC1 vs PC2, colored by true 'charges_cat'
plt.figure(figsize=(6,5))
for cat in np.unique(y_labeled):
    mask = (y_labeled == cat)
    plt.scatter(
        X_pca_labeled[mask, 0],
        X_pca_labeled[mask, 1],
        label=cat,
        alpha=0.6
    )
plt.legend(title="Charges cat")
plt.xlabel("PC1"), plt.ylabel("PC2")
plt.title("PCA of Labeled Training Set")
plt.show()
```

reducing to 2D so we can visually inspect how well the categories separate.

### Hyperparameter tuning for the SVM

```python
from sklearn.model_selection import GridSearchCV

param_grid = {'C': [0.1, 1, 10], 'gamma': ['scale', 0.1, 1]}
grid = GridSearchCV(SVC(), param_grid, cv=3, scoring='accuracy')
grid.fit(X_train_scaled, y_train)
print("Best params:", grid.best_params_)
print("Best CV score:", grid.best_score_)
```

searching over C and gamma to see if a tuned SVM improves accuracy.

### Compare K-Means clusters to true categories

```python
# we originally ran KMeans on all 935 training rows:
from sklearn.cluster import KMeans
km = KMeans(n_clusters=4, random_state=42)
clusters = km.fit_predict(X_train_scaled)    # length 935

# but y_train_labeled is only the 647 rows where y_train != -1
mask_labeled = (y_train != -1)

# so filter the cluster array to those same 647 entries
clusters_labeled = clusters[mask_labeled]    # now length 647

# now cross-tabulate
import pandas as pd
ct = pd.crosstab(
    clusters_labeled,
    y_train_labeled,
    rownames=['cluster'],
    colnames=['true_cat']
)
display(ct)
```

seeing which clusters align (or mix) different charge categories.

### Try another semi-supervised algorithm

```python
from sklearn.semi_supervised import LabelPropagation
from sklearn.metrics import accuracy_score

# 1) Initialize and fit on all training data (with –1 for unlabeled)
lp = LabelPropagation(kernel='rbf', gamma=20, max_iter=1000)
lp.fit(X_train_scaled, y_train)

# 2) Predict labels for all training points
y_lp = lp.predict(X_train_scaled)

# 3) Compute accuracy only on the originally labeled subset
mask_labeled = (y_train != -1)
acc_lp = accuracy_score(y_train[mask_labeled], y_lp[mask_labeled])
print("LabelPropagation accuracy on labeled data:", acc_lp)
```

swapping in LabelPropagation to compare against LabelSpreading performance.

### t-SNE visualization of all points

```python
from sklearn.manifold import TSNE

tsne = TSNE(n_components=2, random_state=42)
X_tsne = tsne.fit_transform(X_train_scaled)

plt.figure(figsize=(6,5))
plt.scatter(X_tsne[:,0], X_tsne[:,1], c=(y_train==-1), cmap='coolwarm', alpha=0.6)
plt.title("t-SNE: red=unlabeled, blue=labeled")
plt.xlabel("Dim 1"), plt.ylabel("Dim 2")
plt.show()
```

mapping points to 2D and coloring labeled vs. unlabeled to check whether the missing labels cluster together.
