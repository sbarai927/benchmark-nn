```python
# %% [1] Setup & imports
import os
import joblib
import numpy as np
import pandas as pd
from sklearn.metrics import mean_squared_error, r2_score
from hyperopt import fmin, tpe, hp, Trials, STATUS_OK
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, optimizers, losses, callbacks, metrics

# reproduce
np.random.seed(42)
tf.random.set_seed(42)
```

```python
# %% [2] Load processed splits
# adjust paths if your data lives elsewhere
X_train, y_train = joblib.load("data/processed/train.pkl")
X_val,   y_val   = joblib.load("data/processed/val.pkl")
X_test,  y_test  = joblib.load("data/processed/test.pkl")

# log-transform the targets (just like your best notebook)
y_train_log = np.log1p(y_train)
y_val_log   = np.log1p(y_val)
```

```python
# %% [3] Define Hyperopt search space
space = {
    "n_layers":     hp.choice("n_layers",     [1, 2, 3, 4, 5]),
    "units":        hp.choice("units",        [32, 64, 128, 256]),
    "dropout":      hp.uniform("dropout",     0.0, 0.5),
    "weight_decay": hp.loguniform("weight_decay", np.log(1e-6), np.log(1e-3)),
    "lr":           hp.loguniform("lr",           np.log(1e-4), np.log(1e-2)),
    "batch_size":   hp.choice("batch_size",   [32, 64, 128]),
}
```

```python
# %% [4] Objective function (with timing)
import time
from hyperopt import STATUS_OK

def objective(params):
    """
    Build, compile, and fit an MLP for a given hyperparameter set.
    Returns the best validation log-RMSE and records the trial duration.
    """
    # Unpack hyperparameters
    n_layers     = params["n_layers"]
    units        = params["units"]
    dropout_rate = params["dropout"]
    wd           = params["weight_decay"]
    lr           = params["lr"]
    bs           = params["batch_size"]

    # Build model
    model = keras.Sequential()
    model.add(layers.Input(shape=(X_train.shape[1],)))
    for _ in range(n_layers):
        model.add(
            layers.Dense(
                units,
                activation="relu",
                kernel_regularizer=keras.regularizers.l2(wd)
            )
        )
        model.add(layers.Dropout(dropout_rate))
    model.add(layers.Dense(1, name="output"))

    # Compile
    model.compile(
        optimizer=optimizers.Adam(learning_rate=lr),
        loss=losses.MeanSquaredError(),
        metrics=[metrics.RootMeanSquaredError(name="rmse")],
    )

    # Early stopping
    es = callbacks.EarlyStopping(
        monitor="val_rmse",
        patience=5,
        restore_best_weights=True,
        verbose=0
    )

    # Train & measure duration
    start_time = time.time()
    history = model.fit(
        X_train, y_train_log,
        validation_data=(X_val, y_val_log),
        epochs=50,
        batch_size=bs,
        callbacks=[es],
        verbose=0
    )
    duration = time.time() - start_time

    # Extract best validation log-RMSE
    best_log_rmse = float(np.min(history.history["val_rmse"]))

    # Return loss & timing for Hyperopt
    return {
        "loss":     best_log_rmse,
        "status":   STATUS_OK,
        "duration": duration
    }
```

```python
# %% [5] Run Hyperopt search
max_evals = 30
trials    = Trials()

best = fmin(
    fn=objective,
    space=space,
    algo=tpe.suggest,
    max_evals=max_evals,
    trials=trials,
    rstate=np.random.default_rng(42),
)

# decode best choices for 'choice' parameters
best_params = {
    "n_layers":     [1,2,3,4,5][ best["n_layers"] ],
    "units":        [32,64,128,256][ best["units"] ],
    "dropout":      best["dropout"],
    "weight_decay": best["weight_decay"],
    "lr":           best["lr"],
    "batch_size":   [32,64,128][ best["batch_size"] ],
}

print("→ Best hyperparameters:")
for k,v in best_params.items():
    print(f"   • {k:13s}= {v}")
```

```python
# %% [6] Summarize all trials to a DataFrame & save
results_df = pd.DataFrame({
    "loss":      [t["result"]["loss"]       for t in trials.trials],
    "duration":  [t["result"]["duration"]   for t in trials.trials],
    "n_layers":  [t["misc"]["vals"]["n_layers"][0] for t in trials.trials],
    "units":     [t["misc"]["vals"]["units"][0]    for t in trials.trials],
    "dropout":   [t["misc"]["vals"]["dropout"][0]  for t in trials.trials],
    "weight_decay":[t["misc"]["vals"]["weight_decay"][0] for t in trials.trials],
    "lr":        [t["misc"]["vals"]["lr"][0]        for t in trials.trials],
    "batch_size":[t["misc"]["vals"]["batch_size"][0] for t in trials.trials],
})
results_df.to_csv("data/processed/hyperopt_summary.csv", index=False)
joblib.dump(trials, "models/hyperopt_trials.pkl")
```
