```python
# 1) Preprocessing: log-transform the target
import numpy as np
import joblib

# load your splits (adjust paths as needed)
X_train, y_train = joblib.load("data/processed/train.pkl")
X_val,   y_val   = joblib.load("data/processed/val.pkl")
X_test,  y_test  = joblib.load("data/processed/test.pkl")

# apply log1p to the targets
y_train_log = np.log1p(y_train)
y_val_log   = np.log1p(y_val)
```

```python
# 2) Define a tuner-friendly model builder
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, losses, metrics, optimizers
from keras_tuner import Hyperband, Objective

def build_model(hp):
    model = keras.Sequential()
    model.add(layers.Input(shape=(X_train.shape[1],)))
    
    # tune number of hidden layers (1–5)
    for i in range(hp.Int("n_layers", 1, 5)):
        model.add(
            layers.Dense(
                units=hp.Choice("units", [32, 64, 128, 256]),
                activation="relu",
            )
        )
        # tune dropout 0–0.5
        model.add(layers.Dropout(hp.Float("dropout", 0.0, 0.5, step=0.1)))
    
    # final regression output
    model.add(layers.Dense(1, activation="linear"))
    
    # compile with tunable weight‐decay on Adam
    wd = hp.Float("weight_decay", 1e-6, 1e-3, sampling="log")
    lr = hp.Float("lr", 1e-4, 1e-2, sampling="log")
    opt = optimizers.Adam(learning_rate=lr, decay=wd)
    
    model.compile(
        optimizer=opt,
        loss=losses.MeanSquaredError(),
        metrics=[metrics.RootMeanSquaredError(name="rmse")]
    )
    return model
```

```python
# 3) Set up the Hyperband tuner
tuner = Hyperband(
    build_model,                            # model-builder fn
    objective=Objective("val_rmse", "min"),# ← specify “minimize”
    max_epochs=50,
    factor=3,
    directory="kt_dir",
    project_name="diamond_prices"
)

# early stopping on val_rmse
stop_early = keras.callbacks.EarlyStopping(
    monitor="val_rmse", patience=5, restore_best_weights=True
)
```

```python
# 4) Run the search (with hard-coded batch_size=32)
tuner.search(
    X_train, y_train_log,
    validation_data=(X_val, y_val_log),
    epochs=50,
    batch_size=32,
    callbacks=[stop_early],
    verbose=2
)
```

```python
tuner.results_summary()
```

```python
# 5) Retrieve the best model & hyperparameters
best_hp    = tuner.get_best_hyperparameters(1)[0]
best_model = tuner.hypermodel.build(best_hp)

print("Best hyperparameters:")
for param in ["n_layers","units","dropout","lr","weight_decay"]:
    print(f" • {param:13s}= {best_hp.get(param)}")
```

```python
# 6) Retrain on (train+val) for up to the best epoch with early-stop
#    so that we can then back-transform and evaluate on the true test set.

# Combine train + val (log) for final fit
X_full = np.vstack([X_train, X_val])
y_full_log = np.concatenate([y_train_log, y_val_log])

# we'll early-stop based on val_rmse on the old val set during training
final_stop = keras.callbacks.EarlyStopping(
    monitor="val_rmse", patience=5, restore_best_weights=True
)

history = best_model.fit(
    X_full, y_full_log,
    validation_split=len(X_val)/len(X_full),  # hold out the val set again
    epochs=50,
    batch_size=32,
    callbacks=[final_stop],
    verbose=1
)

# find which epoch had the lowest val_rmse
val_rmses = history.history["val_rmse"]
best_idx   = int(np.argmin(val_rmses))
best_epoch = best_idx + 1
best_val   = val_rmses[best_idx]

print(f"\n→ Best epoch={best_epoch}, val_RMSE(log)={best_val:.2f}")
```

```python
# 7) Evaluate on TEST: predict, back-transform, compute RMSE & R²
from sklearn.metrics import mean_squared_error, r2_score

y_pred_log = best_model.predict(X_test).ravel()
y_pred      = np.expm1(y_pred_log)          # back-transform
rmse_test   = np.sqrt(mean_squared_error(y_test, y_pred))
r2_test     = r2_score(y_test, y_pred)

print(f"Test RMSE    = {rmse_test:.2f}")
print(f"Test R²      = {r2_test:.3f}")
```
