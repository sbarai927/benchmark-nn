```python
import random, os
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from hyperopt       import fmin, tpe, hp, Trials, STATUS_OK
from tensorflow.keras.callbacks import EarlyStopping
from tensorflow.keras import models, layers, optimizers, losses, metrics

from src.model  import train_and_evaluate
from src.config import param_space
```

### Fix random seeds for reproducibility

```python
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
os.environ["PYTHONHASHSEED"] = str(SEED)
```

By fixing all relevant RNG seeds, we guarantee that our TPE trials, data shuffling, and even Python’s hash order remain identical across runs.

### Load preprocessed splits

```python
X_train, y_train = joblib.load("data/processed/train.pkl")
X_val,   y_val   = joblib.load("data/processed/val.pkl")

print(f"Train: {X_train.shape}, {y_train.shape}")
print(f" Val : {X_val.shape},   {y_val.shape}")
```

Our training set has 151 075 examples with 153 features each; the held-out validation set has 21 583 examples. These splits match our 7:1 train/val ratio and ensure that Hyperopt sees a fairly large pool for both fitting and early-stop evaluation.

### Define the search space

```python
space = {
    "n_layers":   hp.choice("n_layers",   param_space["n_layers"]),
    "units":      hp.choice("units",      param_space["units"]),
    "lr":         hp.loguniform("lr",
                                np.log(param_space["lr_min"]),
                                np.log(param_space["lr_max"])),
    "batch_size": hp.choice("batch_size", param_space["batch_sizes"]),
}
```

We mirror the same four-dimensional search we used in Optuna: number of dense layers, neurons per layer, log-uniform learning rate, and categorical batch sizes.

### Wrap train_and_evaluate with pruning / early-stop logic

```python
def objective(params):
    """Train for up to 20 epochs, but stop early if no val-improvement in 5 epochs."""
    # build a small callback list
    es = EarlyStopping(
        monitor="val_root_mean_squared_error",
        patience=5,
        restore_best_weights=True,
        verbose=0,
    )
    # our train_and_evaluate doesn’t accept callbacks, so we monkey-patch:
    rmse = train_and_evaluate(
        params, X_train, y_train, X_val, y_val,
        callbacks=[es]  # assume you update train_and_evaluate signature to accept callbacks
    )
    return {"loss": float(rmse), "status": STATUS_OK}
```

We train for up to 20 epochs but halt any trial that fails to improve its validation RMSE for 5 consecutive epochs.

This early-stop reduces wasted time on very poor configs and ensures that only promising hyperparameter sets use the full budget.

### Run the Hyperopt search

```python
MAX_EVALS = 30

trials = Trials()
best = fmin(
    fn=objective,
    space=space,
    algo=tpe.suggest,
    max_evals=MAX_EVALS,
    trials=trials,
    rstate=np.random.default_rng(SEED),
)
print("Raw best:", best)
```

We ran 30 trials of our MLP objective under the TPE algorithm, with reproducible sampling (seed 42). After roughly 101 s per trial, Hyperopt returns the indices it sampled—so batch_size=2 means the third element of our [16, 32, 64] list (i.e. 64), and units=0 means the first choice [32, 64, 128] (i.e. 32).

The top configuration was a medium‐depth network (2 hidden layers) of moderate width (32 neurons), trained with a fairly high learning rate (≈4.8 × 10⁻³) and large batch size (64). This combination drove the validation RMSE down to ≈6 881.5, a solid improvement over our baseline.

### Decode categorical choices back to actual values

```python
best_params = {
    "n_layers":   param_space["n_layers"][best["n_layers"]],
    "units":      param_space["units"][best["units"]],
    "lr":         best["lr"],
    "batch_size": param_space["batch_sizes"][best["batch_size"]],
}
print("Decoded best params:", best_params)
```

### Plot RMSE progression over trials

```python
losses = [t["result"]["loss"] for t in trials.trials]
plt.figure(figsize=(6,4))
plt.plot(losses, marker="o")
plt.title("Hyperopt TPE Tuning Progress (Trial 1)")
plt.xlabel("Trial")
plt.ylabel("Validation RMSE")
plt.grid(alpha=0.3)
plt.show()
```

Early trials hovered around 7 700–8 500 RMSE, showing that random initial configurations were already decent.

Spikes (e.g. trial 6 at ~12 700) highlight very poor hyperparameters—exactly the kind we want TPE to avoid in later trials.

Steady improvement kicks in after trial 10, with several runs dipping below 7 000 RMSE.

The best of the 30 ultimately reached ≈6 881, confirming that TPE successfully zeroed in on a well‐balanced architecture and training schedule.

```python
# Save the raw Trials object
joblib.dump(trials, "data/processed/hyperopt_trials_round1.pkl")

# Flatten into a DataFrame and export
records = []
for t in trials.trials:
    # pull out the sampled hyperparams
    vals = t["misc"]["vals"]
    # convert lists to scalars
    rec = {k: v[0] for k, v in vals.items()}
    # record the loss
    rec["loss"] = t["result"]["loss"]
    records.append(rec)

df1 = pd.DataFrame(records)
df1.to_csv("data/processed/hyperopt_search_round1.csv", index=False)

print("Saved round-1 trials to hyperopt_trials_round1.pkl and hyperopt_search_round1.csv")
```

### Define a tighter search space around the previous winner

```python
refined_space = {
    # keep 1 or 2 layers only
    "n_layers": hp.choice("n_layers", [1, 2]),
    # restrict units to [16, 32, 64]
    "units":    hp.choice("units",    [16, 32, 64]),
    # shrink learning‐rate range to [5e-4, 1e-2]
    "lr":       hp.loguniform("lr", np.log(5e-4), np.log(1e-2)),
    # smaller batch sizes only
    "batch_size": hp.choice("batch_size", [8, 16, 32]),
}
```

Shallow nets (1–2 layers) reached the best RMSEs.

Width of 32–64 units was optimal.

Learning rates between 5×10⁻⁴ and 1×10⁻² converged under early stopping.

Small batch sizes (≤32) gave most stable updates.
This “zoom in” should let TPE focus its budget on the most promising region.

### Wrap train_and_evaluate into a Hyperopt objective

```python
def hyperopt_objective(params):
    """
    Returns {'loss': <val_rmse>, 'status': STATUS_OK}.
    Rely on your model.py's EarlyStopping callback
    to prune unpromising runs early.
    """
    # train_and_evaluate returns final validation RMSE
    rmse = train_and_evaluate(params, X_train, y_train, X_val, y_val)
    return {"loss": rmse, "status": STATUS_OK}
```

We remove the explicit ES callback here since our train_and_evaluate already applies it internally. Each trial still halts after 5 non-improving epochs, saving time on bad configs.

### Run the TPE search

```python
MAX_EVALS = 60
trials   = Trials()
best     = fmin(
    fn=hyperopt_objective,
    space=refined_space,
    algo=tpe.suggest,
    max_evals=MAX_EVALS,
    trials=trials,
    rstate=np.random.default_rng(42)  # for reproducibility
)
```

We allotted 40 trials to the TPE sampler (with a fixed RNG seed of 42 for reproducibility). Over the course of roughly 4 hours 22 minutes, Hyperopt evaluated the objective across increasingly promising configurations. The very best validation RMSE it observed was ≈ 6 016.70

### Decode the categorical indices back to actual values

```python
decoded_best = {
    "n_layers":   [1, 2][best["n_layers"]],
    "units":      [16, 32, 64][best["units"]],
    "lr":         best["lr"],
    "batch_size": [8, 16, 32][best["batch_size"]],
}
print("🔍 Refined best params:", decoded_best)
```

The top‐performing model used 2 hidden layers of 64 units each, a learning rate ≈ 9.9 × 10⁻³, and a batch size of 8.

### Plot the RMSE progression over trials

```python
losses = [t["result"]["loss"] for t in trials.trials]
plt.figure(figsize=(6,4))
plt.plot(losses, marker="o")
plt.title("Hyperopt TPE (refined) Tuning Progress (Trial 2)")
plt.xlabel("Trial")
plt.ylabel("Validation RMSE")
plt.grid(alpha=0.3)
plt.show()
```

Initial Exploration (Trials 0–20)

Wide swings in validation RMSE—from as low as ~6 000 up to >18 000—reflect Hyperopt probing very different corners of the search space.

The occasional low‐error outliers (e.g. trial 5 at ~6 000) show that even early on some promising configurations emerge.

Focused Exploitation (Trials 21–40)

After about 20 trials, RMSE values tighten into the 7 000–8 000 range, indicating Hyperopt has converged onto a region of the hyperparameter space that yields consistently good performance.

Variance here is small (roughly ±1 000), suggesting robust improvement and fewer “wild” candidates being tried.

Late‐Stage Refinement (Trials 41–60+)

A few spikes back up toward ~16 000–17 000 indicate occasional exploratory jumps, but the majority of trials remain clustered between 7 500 and 8 500.

This final plateau shows the algorithm is fine‐tuning around its best‐found hyperparameters, with diminishing returns in RMSE gains.

```python
joblib.dump(trials, "data/processed/hyperopt_trials_round2.pkl")

# build a DataFrame of the hyperparams & losses
records = []
for t in trials.trials:
    rec = t["misc"]["vals"]
    rec = {k: v[0] for k,v in rec.items()}  # extract scalar from single-element lists
    rec["loss"] = t["result"]["loss"]
    records.append(rec)

df = pd.DataFrame(records)
df.to_csv("data/processed/hyperopt_search_round2.csv", index=False)
print("Saved hyperopt_search-round2.csv and hyperopt_trials_round2.pkl")
```

```python
best = decoded_best
n_layers, units, lr, batch_size = best["n_layers"], best["units"], best["lr"], best["batch_size"]

# rebuild
hp_model = models.Sequential()
hp_model.add(layers.Input(shape=(X_train.shape[1],)))
for _ in range(n_layers):
    hp_model.add(layers.Dense(units, activation="relu"))
hp_model.add(layers.Dense(1))
hp_model.compile(
    optimizer=optimizers.Adam(learning_rate=lr),
    loss=losses.MeanSquaredError(),
    metrics=[metrics.RootMeanSquaredError()]
)

# retrain 
hp_model.fit(
    np.vstack([X_train, X_val]),
    np.concatenate([y_train, y_val]),
    epochs=20,
    batch_size=batch_size,
    verbose=1
)

# save it
hp_model.save("models/hyperopt_tuned.keras")
print("✅ Saved Hyperopt-tuned model to models/hyperopt_tuned.keras")
```

Epochs 1–7: RMSE plummets from ~15 500 down to ~6 900, as the network quickly captures the bulk of the structure in the data.

Epochs 8–14: Improvements continue but at a slower clip, dipping from ~9 056 to ~6 572—each additional epoch yields progressively smaller RMSE gains.

Epochs 15–20: The curve begins to flatten out around ~5 500 → ~4 930 by the end of training, indicating diminishing returns and that the model is approaching its capacity on the training set under these hyperparameters.
