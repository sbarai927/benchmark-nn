# Ensemble Learning: MLP + Gradient Boosting

### Imports & Data Loading

```python
import time
import joblib
import numpy as np
import pandas as pd
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_squared_error
import matplotlib.pyplot as plt
from tensorflow.keras.models import load_model
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
sns.set_style("whitegrid")
```

```python
# load the same param_space was used in NB 4
from src.config import param_space  
# load Trials object
import joblib
trials = joblib.load("data/processed/hyperopt_trials_round2.pkl")

# find the best trial index
best_idx = np.argmin([t["result"]["loss"] for t in trials.trials])
best = trials.trials[best_idx]["misc"]["vals"]

# decode the raw indices back into real values
decoded_best = {
    "n_layers": param_space["n_layers"][ best["n_layers"][0] ],
    "units":    param_space["units"][    best["units"][0]    ],
    "lr":       trials.trials[best_idx]["result"]["loss"],  # or directly best["lr"][0]
    "batch_size": param_space["batch_sizes"][ best["batch_size"][0] ]
}
print("🔍 Decoded best params:", decoded_best)
```

```python
from tensorflow.keras import models, layers, optimizers, losses, metrics

def build_mlp(n_layers, units, lr, batch_size):
    """
    Returns a compiled MLP model (untrained).
    """
    model = models.Sequential()
    # input layer
    model.add(layers.Input(shape=(X_train.shape[1],)))
    # hidden dense layers
    for _ in range(n_layers):
        model.add(layers.Dense(units, activation="relu"))
    # output layer
    model.add(layers.Dense(1))
    # compile
    model.compile(
        optimizer=optimizers.Adam(learning_rate=lr),
        loss=losses.MeanSquaredError(),
        metrics=[metrics.RootMeanSquaredError()]
    )
    return model
```

```python
# Reload Optuna study (from Notebook 3)
study         = joblib.load("data/processed/optuna_study.pkl")
best_optuna   = study.best_params
```

```python
X_train, y_train = joblib.load("data/processed/train.pkl")
X_val,   y_val   = joblib.load("data/processed/val.pkl")
X_test,  y_test  = joblib.load("data/processed/test.pkl")
```

### Load the tuned MLP

```python
mlp_base   = load_model("models/baseline-model.keras")
mlp_optuna = load_model("models/optuna-tuned.keras")
mlp_hyper  = load_model("models/hyperopt-tuned.keras")
```

### (Re)build & time each of the four models on the *training* set

```python
# GB Regressor
gb = GradientBoostingRegressor(n_estimators=200, learning_rate=0.05, max_depth=4, random_state=42)
t0 = time.time()
gb.fit(X_train, y_train)
gb_time = time.time() - t0
```

```python
# MLP Baseline
t0 = time.time()
mlp_base.fit(X_train, y_train,
             epochs=20,     
             batch_size=32, 
             verbose=0)
base_train_time = time.time() - t0
```

```python
# MLP Optuna
best_optuna = study.best_params 
mlp_opt = build_mlp(**best_optuna)  
t0 = time.time()
mlp_opt.fit(X_train, y_train,
            epochs=50,    
            batch_size=best_optuna["batch_size"],
            verbose=0)
optuna_train_time = time.time() - t0
```

```python
# MLP Hyperopt
decoded = decoded_best  # Hyperopt decoded best
mlp_hyp = build_mlp(**decoded)
t0 = time.time()
mlp_hyp.fit(X_train, y_train,
            epochs=20,     
            batch_size=decoded["batch_size"],
            verbose=0)
hyperopt_train_time = time.time() - t0
```

```python
print(f"Baseline MLP train time:   {base_train_time:.1f}s")
print(f"Optuna-tuned MLP train time: {optuna_train_time:.1f}s")
print(f"Hyperopt-tuned MLP train time: {hyperopt_train_time:.1f}s")
print(f"GB regressor train time:    {gb_time:.1f}s")
```

### Evaluate on test set and compute RMSEs with timing

```python
def eval_model(model, X, y, record_time=False):
    if record_time:
        t0 = time.time()
        preds = model.predict(X)
        tm = time.time() - t0
        return np.sqrt(mean_squared_error(y, preds)), tm
    else:
        preds = model.predict(X)
        return np.sqrt(mean_squared_error(y, preds))
```

```python
rmse_base     = eval_model(mlp_base,     X_test, y_test)
rmse_optuna   = eval_model(mlp_optuna,   X_test, y_test)
rmse_hyperopt = eval_model(mlp_hyper, X_test, y_test)
rmse_gb,      gb_test_time = eval_model(gb, X_test, y_test, record_time=True)
```

### Compare training time vs. error trade-off

```python
labels = ["MLP Baseline", "MLP Optuna", "MLP Hyperopt", "GB Regressor"]
errors = [rmse_base, rmse_optuna, rmse_hyperopt, rmse_gb]
times  = [base_train_time, optuna_train_time, hyperopt_train_time, gb_time]

fig, ax = plt.subplots(figsize=(7,5))
scatter = ax.scatter(times, errors, s=120, edgecolor='black', alpha=0.85)

# annotate each point with its label and exact values
for lbl, t, e in zip(labels, times, errors):
    ax.text(t, e, f"  {lbl}\n  {t:.1f}s, {e:.0f}",
            va='center', ha='left', fontsize=10, weight='bold')

# axis labels & title
ax.set_xlabel("Training time (s)", fontsize=12, weight='bold')
ax.set_ylabel("Test RMSE", fontsize=12, weight='bold')
ax.set_title("Error vs. Training Time Trade-off", fontsize=14, weight='bold')

# enable only y-grid lines for readability
ax.grid(axis='y', linestyle='--', linewidth=0.6, alpha=0.7)

plt.tight_layout()
plt.show()
```

MLP Baseline
Fastest training (∼30 s) but relatively high error (RMSE ≈ 10 600).
Good “out-of-the-box” reference, but clearly under-fitting compared to tuned models.

Optuna-tuned MLP
Small overhead over baseline (38 s), yet worse validation error (RMSE ≈ 24 600).
Indicates that the particular Optuna run over-explored unhelpful regions or insufficient pruning/early-stop logic.

Hyperopt-tuned MLP
Best accuracy of all four (RMSE ≈ 6 400), at the cost of moderate training time (63 s).
Represents the optimal sweet-spot: a shallow but well-configured network converges quickly under pruning, yielding ~40% lower error than the baseline.

Gradient Boosting Regressor
Longest training (121 s) but a competitive RMSE of ≈ 7 700.
Although slower to train, it remains a strong non-neural baseline—only ~20% worse than our best MLP and ~40% better than the naive MLP.

### Remarks

Accuracy vs. Speed: Hyperopt’s TPE search yields the lowest RMSE, halving error relative to the baseline, at ~2× the compute time.

Non-Neural Alternative: GB Regressor, while the slowest, gives reasonably low error without any neural-network tuning.

Poor Runs: The Optuna-tuned MLP underperforms here—suggesting its search space or pruning settings need revisiting.

Recommendation: For this Kaggle Diamonds regression, our Hyperopt-tuned MLP is the best compromise: sub-7 000 RMSE in under a minute of training.
